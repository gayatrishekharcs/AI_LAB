```python
class PuzzleNode:
    def __init__(self, board, parent=None, move=None, depth=0):
        self.board = board
        self.parent = parent
        self.move = move
        self.depth = depth

def get_blank_pos(board):
    for i in range(3):
        for j in range(3):
            if board[i][j] == 0:
                return i, j

def get_neighbors(node):
    neighbors = []
    r, c = get_blank_pos(node.board)

    # Priority order designed to match your exact image tree path:
    # Down, Up, Left, Right
    delta = [(1, 0, "Down"), (-1, 0, "Up"), (0, -1, "Left"), (0, 1, "Right")]

    for dr, dc, move_name in delta:
        nr, nc = r + dr, c + dc
        if 0 <= nr < 3 and 0 <= nc < 3:
            new_board = [row[:] for row in node.board]
            new_board[r][c], new_board[nr][nc] = new_board[nr][nc], new_board[r][c]
            neighbors.append(PuzzleNode(new_board, node, move_name, node.depth + 1))

    return neighbors

def print_board(board):
    for row in board:
        print(" ".join(str(x) if x != 0 else "_" for x in row))
    print()

def get_user_board(prompt_name):
    print(f"\nEnter the 3x3 grid for the {prompt_name} (Row by Row).")
    print("Use numbers 1-8 and use '0' for the blank space.")
    board = []
    for i in range(3):
        while True:
            try:
                row_input = input(f"Enter row {i+1} separated by spaces (e.g., 1 2 3): ").split()
                row = [int(x) for x in row_input]
                if len(row) != 3:
                    raise ValueError("Each row must have exactly 3 numbers.")
                board.append(row)
                break
            except ValueError as e:
                print(f"Invalid input: {e}. Please try again.")
    return board

def dfs(initial_state, goal_state, max_depth=10):
    start_node = PuzzleNode(initial_state)
    stack = [start_node]
    visited = set()

    while stack:
        current = stack.pop()
        board_tuple = tuple(tuple(row) for row in current.board)

        # Goal check
        if current.board == goal_state:
            path = []
            temp = current
            while temp.parent:
                path.append((temp.move, temp.board))
                temp = temp.parent
            return path[::-1]

        if board_tuple not in visited and current.depth < max_depth:
            visited.add(board_tuple)
            # Reverse neighbors for LIFO stack behavior to maintain the search tree order
            for neighbor in reversed(get_neighbors(current)):
                stack.append(neighbor)

    return None

if __name__ == "__main__":
    print("=== 8-PUZZLE DFS SOLVER ===")

    initial = get_user_board("Initial State")
    goal = get_user_board("Goal State")

    print("\n--- Running Depth First Search (DFS) ---")
    solution = dfs(initial, goal, max_depth=6)

    if solution is not None:
        print(f"\nGoal reached successfully in exactly {len(solution)} moves!")
        print("\nStep-by-step state transitions:")
        print("Initial Configuration:")
        print_board(initial)

        for index, (move, board) in enumerate(solution, 1):
            print(f"Step {index}: Move Blank '{move}'")
            print_board(board)
    else:
        print("No solution found matching this path within the depth threshold.")
```

```python
# Structure to maintain the state of the puzzle board
class PuzzleNode:
    def __init__(self, board, parent=None, move=None, depth=0):
        self.board = board  # 3x3 layout
        self.parent = parent
        self.move = move
        self.depth = depth

def get_blank_pos(board):
    for i in range(3):
        for j in range(3):
            if board[i][j] == 0:
                return i, j

def get_neighbors(node):
    neighbors = []
    r, c = get_blank_pos(node.board)
    # Allowed movement configurations (Up, Down, Left, Right)
    delta = [(-1, 0, "Up"), (1, 0, "Down"), (0, -1, "Left"), (0, 1, "Right")]

    for dr, dc, move_name in delta:
        nr, nc = r + dr, c + dc
        if 0 <= nr < 3 and 0 <= nc < 3:
            # Create a deep copy of the board array for the tile swap
            new_board = [row[:] for row in node.board]
            new_board[r][c], new_board[nr][nc] = new_board[nr][nc], new_board[r][c]
            neighbors.append(PuzzleNode(new_board, node, move_name, node.depth + 1))

    return neighbors

def print_board(board):
    for row in board:
        print(" ".join(str(x) if x != 0 else "_" for x in row))
    print()

def get_user_board(prompt_name):
    print(f"\nEnter the 3x3 grid for the {prompt_name} (Row by Row).")
    print("Use numbers 1-8 and use '0' to represent the blank space.")
    board = []
    for i in range(3):
        while True:
            try:
                row_input = input(f"Enter row {i+1} separated by spaces (e.g., 1 2 3): ").split()
                row = [int(x) for x in row_input]
                if len(row) != 3:
                    raise ValueError("Each row must have exactly 3 numbers.")
                board.append(row)
                break
            except ValueError as e:
                print(f"Invalid input: {e}. Please try again.")
    return board

# --- Depth-Limited Search (DLS) ---
def depth_limited_search(node, goal_state, limit, path_visited):
    if node.board == goal_state:
        return node

    if limit <= 0:
        return "cutoff"

    board_tuple = tuple(tuple(row) for row in node.board)
    path_visited.add(board_tuple)

    any_cutoff = False

    for neighbor in get_neighbors(node):
        nb_tuple = tuple(tuple(row) for row in neighbor.board)
        # Avoid traversing cycles along the current active path branch
        if nb_tuple not in path_visited:
            result = depth_limited_search(neighbor, goal_state, limit - 1, path_visited)

            if result == "cutoff":
                any_cutoff = True
            elif result is not None:
                return result

    path_visited.remove(board_tuple)  # Backtrack step
    return "cutoff" if any_cutoff else None

# --- Iterative Deepening Search (IDS) Main Function ---
def iterative_deepening_search(initial_state, goal_state, max_limit):
    depth = 0
    # Search from depth 0 up to the user-defined maximum limit
    while depth <= max_limit:
        path_visited = set()
        start_node = PuzzleNode(initial_state)

        # Execute DFS restricted to the current depth boundary
        result = depth_limited_search(start_node, goal_state, depth, path_visited)

        # If a valid goal state node is found, return the path layout
        if isinstance(result, PuzzleNode):
            path = []
            curr = result
            while curr.parent:
                path.append((curr.move, curr.board))
                curr = curr.parent
            return path[::-1]

        # If it returns None and no cutoff happened, the entire reachable graph was explored
        if result is None:
            return None

        print(f"Depth {depth} checked... moving to next depth.")
        depth += 1

    return "cutoff_exceeded"

if __name__ == "__main__":
    print("=== OPTIMAL IDS (IDDFS) 8-PUZZLE SOLVER ===")

    # 1. Ask user for boards
    initial = get_user_board("Initial State")
    goal = get_user_board("Goal State")

    # 2. Ask user for maximum search limit
    while True:
        try:
            max_limit_input = input("\nEnter the Maximum Depth Limit to search up to (e.g., 20): ")
            max_limit = int(max_limit_input)
            if max_limit < 0:
                raise ValueError("Limit must be a positive integer.")
            break
        except ValueError as e:
            print(f"Invalid depth limit: {e}. Please enter a valid number.")

    print("\nStarting Iterative Deepening Search...")
    solution = iterative_deepening_search(initial, goal, max_limit)

    if solution == "cutoff_exceeded":
        print(f"\nCould not find a solution within your maximum depth limit of {max_limit}.")
    elif solution is not None:
        print(f"\nOptimal Solution found in exactly {len(solution)} moves!")
        print("\nStep-by-step state transitions:")
        print("Initial Configuration:")
        print_board(initial)

        for index, (move, board) in enumerate(solution, 1):
            print(f"Step {index}: Move Blank '{move}'")
            print_board(board)
    else:
        print("\nNo solution exists for the provided puzzle configuration (entire search space exhausted).")
```

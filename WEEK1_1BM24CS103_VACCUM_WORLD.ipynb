```python
# Vacuum Cleaner Agent
# Two-room environment: Room A and Room B

# Taking input from the user
room_A = input("Enter the status of Room A (Dirty/Clean): ").capitalize()
room_B = input("Enter the status of Room B (Dirty/Clean): ").capitalize()
position = input("Enter the initial position of vacuum (A/B): ").upper()

print("\nInitial State:")
print("Room A:", room_A)
print("Room B:", room_B)
print("Vacuum Position: Room", position)

# Vacuum cleaner agent
while room_A == "Dirty" or room_B == "Dirty":

    if position == "A":

        if room_A == "Dirty":
            print("\nVacuum is in Room A")
            print("Room A is Dirty -> Sucking dirt...")
            room_A = "Clean"
            print("Room A is now Clean")

        else:
            print("\nRoom A is Clean -> Moving Right to Room B")
            position = "B"

    elif position == "B":

        if room_B == "Dirty":
            print("\nVacuum is in Room B")
            print("Room B is Dirty -> Sucking dirt...")
            room_B = "Clean"
            print("Room B is now Clean")

        else:
            print("\nRoom B is Clean -> Moving Left to Room A")
            position = "A"

# Goal state reached
print("\nFinal State:")
print("Room A:", room_A)
print("Room B:", room_B)
print("Vacuum Position: Room", position)
print("\nGoal State Reached: Both rooms are Clean!")
```

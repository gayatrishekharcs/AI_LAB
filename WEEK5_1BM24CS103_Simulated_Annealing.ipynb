```python
import math
import random

def calculate_cost(board):
    cost = 0
    n = len(board)
    for i in range(n):
        for j in range(i + 1, n):
            if board[i] == board[j] or abs(board[i] - board[j]) == abs(i - j):
                cost += 1
    return cost

def simulated_annealing(initial_board, initial_temp, cooling_rate, max_steps=100):
    current_board = list(initial_board)
    current_cost = calculate_cost(current_board)
    T = initial_temp

    print("Initial Configuration")
    print(f"Current Board: {current_board} | Cost: {current_cost} | Temp: {T}\n")

    step = 1
    while T > 0.001 and current_cost > 0 and step <= max_steps:
        print(f"--- Step {step} ---")

        col = random.randint(0, 7)
        current_row = current_board[col]
        possible_rows = [r for r in range(8) if r != current_row]
        next_row = random.choice(possible_rows)

        next_board = list(current_board)
        next_board[col] = next_row
        next_cost = calculate_cost(next_board)

        delta_E = current_cost - next_cost

        print(f"Action: Pick Column {col}, move queen from row {current_row} to row {next_row}")
        print(f"Proposed Board: {next_board} | Proposed Cost: {next_cost}")

        if delta_E > 0:
            current_board = next_board
            current_cost = next_cost
            print(f"Decision: ACCEPT (Better move, delta_E = {delta_E})")
        else:
            p = math.exp(delta_E / T)
            r = random.random()
            print(f"Acceptance Calculation: P = e^({delta_E} / {T:.4f}) = {p:.4f} | Random Roll: R = {r:.4f}")
            if r < p:
                current_board = next_board
                current_cost = next_cost
                print("Decision: ACCEPT (Worse/equal move accepted probabilistically)")
            else:
                print("Decision: REJECT (Revert to previous board)")

        T *= cooling_rate
        print(f"Current Board: {current_board} | Cost: {current_cost} | Updated Temp: {T:.4f}\n")
        step += 1

    if current_cost == 0:
        print(f"Success! Solution Found: {current_board} with cost {current_cost}")
    else:
        print(f"Terminated. Final state: {current_board} with cost {current_cost}")

    return current_board

initial_board = [0, 1, 2, 3, 4, 5, 6, 7]
initial_temp = 10.0
cooling_rate = 0.90

final_board = simulated_annealing(initial_board, initial_temp, cooling_rate)
```

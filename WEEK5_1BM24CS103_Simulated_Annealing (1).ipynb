```python
import random
import math

def calculate_cost(state):
    conflicts = 0
    n = len(state)
    for i in range(n):
        for j in range(i + 1, n):
            if state[i] == state[j]:
                conflicts += 1
            elif abs(state[i] - state[j]) == abs(i - j):
                conflicts += 1
    return conflicts

def get_random_neighbour(state):
    neighbour = list(state)
    col = random.randint(0, 7)
    current_row = neighbour[col]

    new_row = random.randint(0, 7)
    while new_row == current_row:
        new_row = random.randint(0, 7)

    neighbour[col] = new_row
    return neighbour

def simulated_annealing():
    current_state = [random.randint(0, 7) for _ in range(8)]
    current_cost = calculate_cost(current_state)

    T = 100.0
    cooling_rate = 0.995
    threshold = 0.001
    iteration = 0

    while T > threshold:
        if current_cost == 0:
            break

        next_state = get_random_neighbour(current_state)
        next_cost = calculate_cost(next_state)

        delta_E = current_cost - next_cost

        if delta_E > 0:
            current_state = next_state
            current_cost = next_cost
        else:
            p = math.exp(delta_E / T)
            if random.random() < p:
                current_state = next_state
                current_cost = next_cost

        T *= cooling_rate
        iteration += 1

    return current_state, current_cost, iteration

def print_board(state):
    for row in range(8):
        line = ""
        for col in range(8):
            if state[col] == row:
                line += " Q "
            else:
                line += " . "
        print(line)

if __name__ == "__main__":
    print("Running Simulated Annealing for 8-Queens...")

    solution_found = False
    attempts = 0

    while not solution_found and attempts < 100:
        attempts += 1
        final_state, final_cost, total_iters = simulated_annealing()
        if final_cost == 0:
            solution_found = True
            print(f"\nSuccess! Solution found on attempt {attempts} after {total_iters} iterations.")
            print(f"Final State (Rows per column): {final_state}")
            print(f"Conflicts: {final_cost}\n")
            print_board(final_state)
            break

    if not solution_found:
        print("\nCould not find a perfect solution within maximum attempts. Try running again!")
```

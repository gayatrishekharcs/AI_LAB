```python
import itertools

def calculate_cost(state):
    cost = 0
    n = len(state)
    for i in range(n):
        for j in range(i + 1, n):
            if abs(state[i] - state[j]) == abs(i - j):
                cost += 1
    return cost

def get_neighbors_with_costs(state):
    neighbors = []
    n = len(state)
    for i in range(n):
        for j in range(i + 1, n):
            neighbor = list(state)
            neighbor[i], neighbor[j] = neighbor[j], neighbor[i]
            cost = calculate_cost(neighbor)
            neighbors.append((neighbor, cost, (i, j)))
    return neighbors

def format_state(state):
    return ", ".join(f"x_{i} = {val}" for i, val in enumerate(state))

def hill_climbing(initial_state):
    current_state = list(initial_state)
    current_cost = calculate_cost(current_state)
    step = 1
    while True:
        print(f"\n--- Step {step} ---")
        print(f"Current State: {format_state(current_state)} (cost is {current_cost})")
        print("Neighbours:")

        neighbors = get_neighbors_with_costs(current_state)
        best_neighbor = None
        best_cost = float('inf')
        best_pair = None

        for neighbor, cost, pair in neighbors:
            is_chosen = False
            if cost < best_cost:
                best_cost = cost
                best_neighbor = neighbor
                best_pair = pair
                is_chosen = True

            status_str = "Choose this neighbour." if (cost < current_cost and is_chosen) else ""
            print(f"  • {format_state(neighbor)} cost is {cost}. {status_str}".strip())

        if best_cost >= current_cost:
            print("\nTerminated: No strictly better neighbor found.")
            break

        current_state = best_neighbor
        current_cost = best_cost
        step += 1

    print(f"\nFinal Solution State: {format_state(current_state)} with cost {current_cost}")
    return current_state

initial_state = [3, 1, 2, 0]
final_state = hill_climbing(initial_state)
```

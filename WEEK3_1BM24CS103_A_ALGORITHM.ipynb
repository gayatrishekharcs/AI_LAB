```python
import heapq

def get_blank_position(state):
    for r in range(3):
        for c in range(3):
            if state[r][c] == 0:
                return r, c

def calculate_heuristic(state, goal_state):
    misplaced = 0
    for r in range(3):
        for c in range(3):
            if state[r][c] != 0 and state[r][c] != goal_state[r][c]:
                misplaced += 1
    return misplaced

def get_neighbors(state):
    neighbors = []
    r, c = get_blank_position(state)

    moves = {
        'L': (0, -1),
        'U': (-1, 0),
        'R': (0, 1),
        'D': (1, 0)
    }

    for move_name, (dr, dc) in moves.items():
        nr, nc = r + dr, c + dc
        if 0 <= nr < 3 and 0 <= nc < 3:
            new_state = [list(row) for row in state]
            new_state[r][c], new_state[nr][nc] = new_state[nr][nc], new_state[r][c]
            neighbors.append((tuple(tuple(row) for row in new_state), move_name))

    return neighbors

def print_matrix(state):
    for row in state:
        print("  " + " ".join(str(x) if x != 0 else "_" for x in row))

def a_star_search(initial_state, goal_state):
    h_initial = calculate_heuristic(initial_state, goal_state)
    frontier = [(h_initial, 0, initial_state, [])]
    visited = set()
    step = 0

    print("Initial State:")
    print_matrix(initial_state)
    print(f"g = 0, h = {h_initial}, f = {h_initial}\n")

    while frontier:
        f, g, current, path = heapq.heappop(frontier)

        if current in visited:
            continue
        visited.add(current)

        if current == goal_state:
            print("--- Goal State Reached! ---")
            print(f"Total moves: {len(path)}")
            print(f"Path sequence: {' -> '.join(path)}")
            return

        step += 1
        print(f"--- Expanding Node (Step {step}) ---")
        print_matrix(current)
        print(f"Current node values: g = {g}, h = {f - g}, f = {f}")
        print("Generated Successors:")

        for neighbor, move in get_neighbors(current):
            if neighbor not in visited:
                g_new = g + 1
                h_new = calculate_heuristic(neighbor, goal_state)
                f_new = g_new + h_new
                heapq.heappush(frontier, (f_new, g_new, neighbor, path + [move]))
                print(f"  Move {move}:")
                print_matrix(neighbor)
                print(f"    g = {g_new}, h = {h_new}, f = {f_new}")
        print()

initial = (
    (2, 8, 3),
    (1, 6, 4),
    (7, 0, 5)
)

goal = (
    (1, 2, 3),
    (8, 0, 4),
    (7, 6, 5)
)

a_star_search(initial, goal)
```

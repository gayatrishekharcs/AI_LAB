```python
import math
import random

def print_board(board):
    print("\n")
    for i in range(3):
        print(f" {board[i*3]} | {board[i*3+1]} | {board[i*3+2]} ")
        if i < 2:
            print("---|---|---")
    print("\n")

def check_win(board, player):
    win_conditions = [
        [0, 1, 2], [3, 4, 5], [6, 7, 8], # Rows
        [0, 3, 6], [1, 4, 7], [2, 5, 8], # Columns
        [0, 4, 8], [2, 4, 6]            # Diagonals
    ]
    for cond in win_conditions:
        if board[cond[0]] == board[cond[1]] == board[cond[2]] == player:
            return True
    return False

def is_board_full(board):
    return all(spot in ['X', 'O'] for spot in board)

def get_computer_move(board, computer_char, human_char):
    for i in range(9):
        if board[i] == ' ':
            board[i] = computer_char
            if check_win(board, computer_char):
                board[i] = ' '
                return i
            board[i] = ' '

    for i in range(9):
        if board[i] == ' ':
            board[i] = human_char
            if check_win(board, human_char):
                board[i] = ' '
                return i
            board[i] = ' '

    if board[4] == ' ':
        return 4

    available_moves = [i for i, spot in enumerate(board) if spot == ' ']
    return random.choice(available_moves) if available_moves else None

def play_tic_tac_toe():
    print("=== Tic-Tac-Toe: Human vs Computer ===")
    print("USN: 1BM24CS103 -AND- NAME: GAYATRI SHEKHAR JIRALI")
    board = [' ' for _ in range(9)]
    human_char = input("Choose your character ('X' or 'O'): ").upper()
    while human_char not in ['X', 'O']:
        human_char = input("Invalid choice! Choose 'X' or 'O': ").upper()

    computer_char = 'O' if human_char == 'X' else 'X'
    turn = 'HUMAN' if human_char == 'X' else 'COMPUTER'

    print(f"You are {human_char}, Computer is {computer_char}.")

    while True:
        print_board(board)
        if turn == 'HUMAN':
            try:
                move_input = int(input("Enter your move position (1-9): "))
                move = move_input - 1
                if 0 <= move <= 8 and board[move] == ' ':
                    board[move] = human_char
                    if check_win(board, human_char):
                        print_board(board)
                        print("Congratulations! You win!")
                        break
                    elif is_board_full(board):
                        print_board(board)
                        print("It's a draw!")
                        break
                    turn = 'COMPUTER'
                else:
                    print("Invalid move! Spot is already taken or out of range. Choose another.")
            except (ValueError, IndexError):
                print("Invalid input! Enter a number between 1 and 9.")
        else:
            print("Computer is making a move...")
            move = get_computer_move(board, computer_char, human_char)
            if move is not None:
                board[move] = computer_char
                if check_win(board, computer_char):
                    print_board(board)
                    print("Computer wins! Better luck next time.")
                    break
                elif is_board_full(board):
                    print_board(board)
                    print("It's a draw!")
                    break
                turn = 'HUMAN'
            else:
                print("No moves left for computer.")
                break

if __name__ == "__main__":
    play_tic_tac_toe()
```
